```python
path = "/content/drive/MyDrive/Library_Usage.csv"
import pandas as pd
df = pd.read_csv(path)
df.head()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="pie",
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="bar",
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="pink",
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="pink",
)
plt.title("Library Circulation Activity by Month")
plt.xlabel("Month")
plt.ylabel("# of Circulations")
plt.show()
```

```python
month.order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"
]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="pink",
)
plt.title("Library Circulation Activity by Month")
plt.xlabel("Month")
plt.ylabel("# of Circulations")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]

df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="aquamarine",
    edgecolor="black"
)

# --------------------------------------------------------------------------
# Play even more with how the chart looks!
# --------------------------------------------------------------------------
plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
df["Circulation Active Month CLEANED"] = df["Circulation Active Month"].fillna('Not Captured')

df.isna().sum()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December", "Not Captured"]

df["Circulation Active Month CLEANED"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="pink",
    edgecolor="purple"
)

# --------------------------------------------------------------------------
# Play even more with how the chart looks!
# --------------------------------------------------------------------------
plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python

df["Age Range"].value_counts().plot(
    kind="bar",
    color="pink",
)
plt.title("Age Range")
plt.xlabel("Age")
plt.ylabel("# of Patrons")
plt.show()
```

```python
df["Age Range"].value_counts().plot(
    kind="density",
    color="pink",
)
plt.title("Age Range")
plt.xlabel("Age")
plt.ylabel("# of Patrons")
plt.show()
```

```python
age_order = [
    "0 to 9 years", "10 to 19 years", "20 to 24 years", "25 to 34 years", "35 to 44 years", "45 to 54 years",
    "55 to 59 years", "60 to 64 years", "65 to 74 years", "75 years and over"]

df["Age Range"].value_counts().reindex(age_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="aquamarine",
    edgecolor="black"
)

# --------------------------------------------------------------------------
# Play even more with how the chart looks!
# --------------------------------------------------------------------------
plt.title("Age Range of Patrons", fontsize=16, weight="bold")
plt.xlabel("Age Range", fontsize=12)
plt.ylabel("Number of Patrons", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
